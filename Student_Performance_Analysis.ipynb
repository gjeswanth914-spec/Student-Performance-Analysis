# Importing Required Libraries

In this step, the required Python libraries for data analysis and visualization are imported.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# Loading the Dataset

The student performance dataset is loaded using Pandas.

```python
df = pd.read_csv('student-mat.csv', sep=';')

print(df.head())
```

# Understanding Dataset Dimensions

The shape of the dataset is checked to identify the number of rows and columns.

```python
print(df.shape)
```

# Checking Missing Values

This step checks whether the dataset contains any missing values.

```python
print(df.isnull().sum())
```

# Checking Duplicate Rows

This step checks for duplicate records in the dataset.

```python
print(df.duplicated().sum())
```

# Calculating Average Final Grade

The average final grade (G3) of students is calculated.

```python
average_grade = df['G3'].mean()

print("Average Final Grade:", average_grade)
```

# Students Scoring Above 15

This step identifies number of  students who scored above 15 in the final grade.

```python
high_scorers = df[df['G3'] > 15]

print("Number of students scoring above 15:", len(high_scorers))
```

# Correlation Analysis

This analysis checks the relationship between study time and final grades.

```python
correlation = df['studytime'].corr(df['G3'])

print("Correlation between Study Time and Final Grade:", correlation)
```

# Gender-wise Performance Analysis

The average performance of male and female students is compared.

```python
gender_performance = df.groupby('sex')['G3'].mean()

print(gender_performance)
```

# Data Visualization

Data visualization helps in understanding patterns, trends, and relationships within the dataset.

## Histogram of Final Grades

This histogram represents the distribution of students final grades (G3).

```python
plt.figure(figsize=(8,5))

plt.hist(df['G3'], bins=10)

plt.title('Distribution of Final Grades')
plt.xlabel('Final Grade')
plt.ylabel('Number of Students')

plt.show()
```

## Scatter Plot: Study Time vs Final Grade

This scatter plot shows the relationship between study time and students' final grades.

```python
plt.figure(figsize=(8,5))

plt.scatter(df['studytime'], df['G3'])

plt.title('Study Time vs Final Grade')
plt.xlabel('Study Time')
plt.ylabel('Final Grade')

plt.show()
```

## Gender-wise Average Final Grade

This bar chart compares the average final grades of male and female students.

```python
plt.figure(figsize=(6,5))

gender_performance.plot(kind='bar')

plt.title('Average Final Grade by Gender')
plt.xlabel('Gender')
plt.ylabel('Average G3 Score')

plt.show()
```
